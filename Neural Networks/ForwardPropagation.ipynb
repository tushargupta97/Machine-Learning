```python
import numpy as np
```

```python
X=np.array([[0,0],[0,1],[1,0],[1,1]])
Y=np.array([[0,0,0,1]]).T
print(X.shape)
print(Y.shape)
```

```python
def sig(z):
    return (1/(1+np.exp(-z)))
```

```python
#no hidden layer weights
weights= 2*np.random.random((2,1))-1
bias=2*np.random.random(1)-1
weights , bias
```

```python
#forward propagation without any hidden layer
output0=X
output=sig(np.dot(output0, weights) + bias)
output
```

```python
wh = 2 * np.random.random((2,2))-1 #hidden weights
bh = 2 * np.random.random((1,2))-1 #hiddent bias
wo= 2 * np.random.random((2,1))-1 #output weights
bo= 2 * np.random.random((1,1))-1 #output bias
```

```python
output0=X
outputHidden=sig(np.dot(output0, wh) + bh)
output=sig(np.dot(outputHidden, wo) + bo)
output
```
